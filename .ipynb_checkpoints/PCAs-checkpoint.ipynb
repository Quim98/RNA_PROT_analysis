```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.metrics import r2_score
from scipy.stats import spearmanr
import numpy as np
import matplotlib.pyplot as plt
import warnings
from multiprocessing import Pool
import pickle
from src.simulate_func import gen_lasso_model,parallel_gen_lasso_model,lasso_model_sample,parallel_sim_lasso
import time
```

```python
# IT CAN DO A SIMULATION either train or test, but the second one always gets stuck
```

```python
alpha_val = 0.4
# Get transcriptomics and proteomics data
df = pd.read_csv("data/whole_dataset.csv").dropna()
# Get data from only 1 RNA and protein dataset in immune cells
df = df.loc[(df["RNA dataset num."]==1)&(df["Prot. count dataset num."]==1)]
df = df.loc[~df["Cell"].isin(["T4.CM_IL2RAlow","T4.EM_IL2RAlow","T4.naive_IL2RAlow"])]
gene_both_datasets = df["Gene"].drop_duplicates()[df["Gene"].drop_duplicates().isin(pd.read_csv("data/tissues_dataset.csv").dropna()["Gene"].drop_duplicates())].values
df = df.loc[df["Gene"].isin(gene_both_datasets)]

num_samples = 1 #50
num_cores = 4 #24

# Only get genes with enough data and that are differnetially expressed
Num_data = [len(df.loc[df["Gene"]==gene].index) for gene in df["Gene"].drop_duplicates()]
var_TPM = [df.loc[df["Gene"]==gene,"Log2 TPM"].var() for gene in df["Gene"].drop_duplicates()[Num_data >= np.median(Num_data)]]
gene_list = df["Gene"].drop_duplicates()[Num_data >= np.median(Num_data)][var_TPM >= np.median(var_TPM)].values
df = df.loc[df["Gene"].isin(gene_list)]
df["Sample"] = np.nan
df["Prot. count predict"] = np.nan

# Separate TPM and protein data into training and test 10 times (80/20 split)
df_TPM = df.pivot(index='Cell', columns='Gene', values='Log2 TPM')
df_prot = df.pivot(index='Cell', columns='Gene', values='Log10 Prot. count').fillna(0)
df_train_TPM = pd.DataFrame(columns=list(df_TPM.columns)+["Cell","Sample"])
df_test_TPM = pd.DataFrame(columns=list(df_TPM.columns)+["Cell","Sample"])
warnings.filterwarnings('ignore')
for i in range(0,num_samples):
    df_train_slice_TPM, df_test_slice_TPM = train_test_split(df_TPM, test_size=0.2, random_state=i)
    df_train_slice_TPM["Cell"] = df_train_slice_TPM.index
    df_test_slice_TPM["Cell"] = df_test_slice_TPM.index
    df_train_slice_TPM["Sample"] = i
    df_test_slice_TPM["Sample"] = i
    df_train_TPM = pd.concat([df_train_TPM, df_train_slice_TPM.fillna(0)], ignore_index=True)
    df_test_TPM = pd.concat([df_test_TPM, df_test_slice_TPM.fillna(0)], ignore_index=True)
warnings.filterwarnings('default')
print("Data sampling done!")

# Generate linear model for all samples
warnings.filterwarnings('ignore')
list_models, df_models = parallel_gen_lasso_model(df_train_TPM, df_prot, gene_list, num_samples, num_cores, alpha_val)
warnings.filterwarnings('default')
df_models.to_csv('results/lasso_model/lasso_immune'+str(alpha_val)+'.tsv.gz', sep='\t', index=False, compression='gzip')
with open('results/lasso_model/lasso_immune'+str(alpha_val)+'.pkl', 'wb') as file:
    pickle.dump(list_models, file)
print("Lasso models done!")
```

```python
# Simulate for the training set of each sample
warnings.filterwarnings('ignore')
df_train = parallel_sim_lasso(df_train_TPM, df, list_models, gene_list, num_samples, num_cores)
# Get metrics for training set
df_train_res = pd.DataFrame(columns=["Sample","r2","MAE","Spearman (TPMxProt)","Spearman (ModelxProt)"])
for i in range(0,num_samples):
    df_train_res.loc[len(df_train_res)] = [i,r2_score(df_train.loc[df_train["Sample"]==i,"Log10 Prot. count"], df_train.loc[df_train["Sample"]==i,"Prot. count predict"]),(df_train.loc[df_train["Sample"]==i,"Log10 Prot. count"]-df_train.loc[df_train["Sample"]==i,"Prot. count predict"]).abs().mean(),spearmanr(df_train.loc[df_train["Sample"]==i,"Log2 TPM"],df_train.loc[df_train["Sample"]==i,"Log10 Prot. count"]).statistic,spearmanr(df_train.loc[df_train["Sample"]==i,"Prot. count predict"],df_train.loc[df_train["Sample"]==i,"Log10 Prot. count"]).statistic]
df_train.to_csv('results/lasso_model/lasso_immune'+str(alpha_val)+'_train.tsv.gz', sep='\t', index=False, compression='gzip')
df_train_res.to_csv('results/lasso_model/lasso_immune'+str(alpha_val)+'_train_metrics.tsv.gz', sep='\t', index=False, compression='gzip')
del(df_train)
del(df_train_res)
del(list_models)
print("Training set simulation done!")
warnings.filterwarnings('default')
```

```python
from copy import deepcopy
```

```python
def lasso_model_sample(list_df):
    df_TPM_sample = list_df[0]
    df_res_sample = list_df[1]
    gene_list = list_df[2]
    sample =  list_df[3]
    with open('results/lasso_model/lasso_immune' + str(alpha_val) + '.pkl', 'rb') as file:
        list_models = pickle.load(file)
    model_dict = list_models[sample]
    del(list_models)
    for gene in gene_list:
        model_gene = deepcopy(model_dict[gene])
        df_TPM_sample_gene = df_TPM_sample.set_index('Cell').reindex(df_res_sample.loc[df_res_sample["Gene"]==gene,"Cell"])
        df_res_sample.loc[df_res_sample["Gene"]==gene,"Prot. count predict"] = model_gene.predict(df_TPM_sample_gene[df_TPM_sample_gene.columns[:-1]].values)
        print("Works")
    return df_res_sample

def parallel_sim_lasso(df_data_TPM, df_melted, model_list, gene_list, num_samples, num_cores):
    param_parallel = []
    print(1)
    for i in range(0,num_samples):
        df_res_sample = df_melted.loc[df_melted["Cell"].isin(df_data_TPM.loc[df_data_TPM["Sample"]==i,"Cell"])]
        df_res_sample["Sample"] = i
        param_parallel.append([df_data_TPM.loc[df_data_TPM["Sample"]==i], df_res_sample, gene_list, i])
    with Pool(processes=num_cores) as pool:
        out_vec = pool.map(lasso_model_sample, param_parallel)
    return pd.concat(out_vec, axis=0)
```
